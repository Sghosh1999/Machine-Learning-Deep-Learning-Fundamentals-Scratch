## Creating a simple Perceptron Model using PyTorch

### Dataset: Lets assume we have a students IQ, CGPA< Independent Variables and Pass/Fail is the dependent varaible

### Generate a Sample Students PAss/Fail Data

```python
## Creating a sample dataset
import pandas as pd
import numpy as np

students_iq = np.random.randint(80,150, size=100)
students_cgpa = np.random.uniform(1.0, 10.0, size=100)
students_pass_fail = np.where(
    (students_iq > 70) & (students_cgpa > 7.5),
    np.random.choice([1, 0], size=100, p=[0.85, 0.15]),  # mostly pass if both conditions met
    np.random.choice([1, 0], size=100, p=[0.3, 0.7])     # mostly fail otherwise
)
students_pass_fail = np.random.choice([0, 1], size=100)  # 0 for Fail, 1 for Pass

students_data = pd.DataFrame({
    'IQ': students_iq,
    'CGPA': students_cgpa,
    'Pass/Fail': students_pass_fail
})
students_data.head()
```

```python
## Creating a sample Perceptron Model using PyTorch
import torch
import torch.nn as nn
import torch.optim as optim

class SimplePerceptron(nn.Module):
    def __init__(self, input_size, output_size):
        super(SimplePerceptron, self).__init__()
        self.linear = nn.Linear(input_size, output_size)
    
    def forward(self, x):
        return torch.sigmoid(self.linear(x))
    
# Initialize the model, loss function, and optimizer
input_size = 2  # IQ and CGPA
output_size = 1  # Pass/Fail
model = SimplePerceptron(input_size, output_size)

criterion = nn.BCELoss()  # Binary Cross Entropy Loss
optimizer = optim.SGD(model.parameters(), lr=0.01)  

# Convert DataFrame to PyTorch tensors
X = torch.tensor(students_data[['IQ', 'CGPA']].values, dtype=torch.float32)
y = torch.tensor(students_data['Pass/Fail'].values, dtype=torch.float32).view(-1, 1)


## Training the model
for epoch in range(100):  # Number of epochs
    model.train()
    optimizer.zero_grad()  # Zero the gradients
    outputs = model(X)  # Forward pass
    loss = criterion(outputs, y)  # Compute the loss
    loss.backward()  # Backward pass

    optimizer.step()  # Update the weights
    if (epoch+1) % 10 == 0:  # Print loss every 10 epochs
        print(f'Epoch [{epoch+1}/100], Loss: {loss.item():.4f}')
```
