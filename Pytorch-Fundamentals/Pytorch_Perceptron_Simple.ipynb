```python
import torch
import torch.nn as nn
from torch.optim import Adam

class MultiLayerPerceptron(nn.Module):
    def __init__(self, input_size, output_size):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(input_size, 128), #Input layer
            nn.ReLU(),
            nn.Linear(128, 64), #Hidden layer | Size (128, 64)
            nn.ReLU(),
            nn.Linear(64, output_size) #Output layer
        )
    
    def forward(self, x):
        return self.model(x)
    

#Inputs
X = torch.randn(10,2, dtype=torch.float32) #10 samples, binary target
y = torch.randint(0,2, (10,), dtype=torch.float32) #Binary target

#Model
mymodel = MultiLayerPerceptron(input_size=2, output_size=1)
#Loss and optimizer
criterion = nn.BCEWithLogitsLoss()
optimizer = Adam(mymodel.parameters(), lr=0.001)
#Training loop
for epoch in range(100):
    mymodel.train()
    optimizer.zero_grad()  # Zero the gradients

    outputs = mymodel(X)  # Forward pass
    loss = criterion(outputs.squeeze(), y)  # Compute loss
    loss.backward()  # Backward pass
    optimizer.step()  # Update weights
    
    if (epoch + 1) % 10 == 0:
        print(f"Epoch [{epoch+1}/100], Loss: {loss.item():.4f}")
```

```python
## Predict for a Unseen sample
sample = torch.tensor([[0.5, 0.5]], dtype=torch.float32)  # Example input
with torch.inference_mode():
    prediction = mymodel(sample)
    predicted_class = torch.sigmoid(prediction).round()  # Apply sigmoid and round to get binary output
    print(f"Predicted class for sample {sample.numpy()} is: {predicted_class.item()}")
```
