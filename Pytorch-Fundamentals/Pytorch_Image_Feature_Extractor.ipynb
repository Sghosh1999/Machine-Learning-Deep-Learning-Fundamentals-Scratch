```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.transforms as transforms
import torchvision.models as models
from PIL import Image
```

```python
# --------------------------
# 1. Load Pretrained ResNet50 without final classifier
# --------------------------
resnet50 = models.resnet50(pretrained=True)

## Remove the Classifier Head
resnet50.fc = nn.Identity()
```

## Image Preprocessing

```python
## Image Preprocessing
preprocess = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])
```

```python
## Image Path
# Set the path to the image file and preprocess it for the model
image_path = 'images/car-image-1.jpg'
image = Image.open(image_path)
image_tensor = preprocess(image).unsqueeze(0)  # Add batch dimension
image_tensor  
```

```python
# --------------------------
# 3. Extract Embeddings
# --------------------------
with torch.no_grad():
    embedding = resnet50(image_tensor)  # Shape: [1, 2048, 1, 1]
    embedding = embedding.view(embedding.size(0), -1)  # Flatten -> [1, 2048]
```

```python
# --------------------------
# 4. Display the Embedding
# --------------------------
print("Embedding shape:", embedding.shape)  # (1, 2048)
print("Embedding vector:", embedding[0][:10])  # Display first 10 values
```
