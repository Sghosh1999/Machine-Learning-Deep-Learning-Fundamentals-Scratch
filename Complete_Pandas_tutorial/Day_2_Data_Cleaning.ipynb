# Day 2: Data Cleaning and Missing Values

Data cleaning is crucial for any data analysis project. Today we'll learn how to handle missing values, duplicates, and transform data.

```python
import pandas as pd
import numpy as np

# Create sample data with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Alice'],
    'Age': [25, np.nan, 35, 28, 32, 45, 25],
    'City': ['New York', 'London', None, 'Paris', 'Berlin', 'Tokyo', 'New York'],
    'Salary': [50000, 60000, 70000, np.nan, 65000, 80000, 50000],
    'Email': ['alice@email.com', 'bob@email.com', 'charlie@email.com', 
              'david@email.com', 'eva@email.com', 'frank@email.com', 'alice@email.com']
}
df = pd.DataFrame(data)
print("Sample DataFrame with missing values:")
print(df)
print()
print("Data types:")
print(df.dtypes)
```

## 1. Identifying Missing Values

```python
# Check for missing values
print("Missing values per column:")
print(df.isnull().sum())
print()

print("Missing values (boolean):")
print(df.isnull())
print()

print("Non-missing values:")
print(df.notnull().sum())
print()

# Info about missing data
print("DataFrame info:")
df.info()
```

## 2. Handling Missing Values

```python
# Drop rows with any missing values
df_dropped_rows = df.dropna()
print("After dropping rows with missing values:")
print(df_dropped_rows)
print(f"Original shape: {df.shape}, New shape: {df_dropped_rows.shape}")
print()

# Drop columns with any missing values
df_dropped_cols = df.dropna(axis=1)
print("After dropping columns with missing values:")
print(df_dropped_cols)
print(f"Original columns: {len(df.columns)}, New columns: {len(df_dropped_cols.columns)}")
print()

# Drop rows where specific columns have missing values
df_specific = df.dropna(subset=['Age', 'Salary'])
print("After dropping rows with missing Age or Salary:")
print(df_specific)
```

## 3. Filling Missing Values

```python
# Fill with a constant value
df_filled = df.fillna(0)
print("Filled with 0:")
print(df_filled)
print()

# Fill with different values for different columns
fill_values = {'Age': df['Age'].mean(), 'City': 'Unknown', 'Salary': df['Salary'].median()}
df_filled_specific = df.fillna(value=fill_values)
print("Filled with specific values:")
print(df_filled_specific)
print()

# Forward fill (use previous value)
df_ffill = df.fillna(method='ffill')
print("Forward fill:")
print(df_ffill)
print()

# Backward fill (use next value)
df_bfill = df.fillna(method='bfill')
print("Backward fill:")
print(df_bfill)
```

## 4. Interpolation

```python
# Linear interpolation for numeric columns
df_interpolated = df.copy()
df_interpolated['Age'] = df_interpolated['Age'].interpolate()
df_interpolated['Salary'] = df_interpolated['Salary'].interpolate()

print("After interpolation:")
print(df_interpolated)
print()

# Different interpolation methods
df_poly = df.copy()
df_poly['Salary'] = df_poly['Salary'].interpolate(method='polynomial', order=2)
print("Polynomial interpolation for Salary:")
print(df_poly[['Name', 'Salary']])
```

## 5. Handling Duplicates

```python
# Check for duplicates
print("Duplicate rows:")
print(df.duplicated())
print()

print("Number of duplicate rows:", df.duplicated().sum())
print()

# Show duplicate rows
print("Actual duplicate rows:")
print(df[df.duplicated()])
print()

# Remove duplicates
df_no_duplicates = df.drop_duplicates()
print("After removing duplicates:")
print(df_no_duplicates)
print(f"Original shape: {df.shape}, New shape: {df_no_duplicates.shape}")
print()

# Remove duplicates based on specific columns
df_no_name_duplicates = df.drop_duplicates(subset=['Name'])
print("After removing duplicates based on Name:")
print(df_no_name_duplicates)
```

## 6. String Operations

```python
# String methods
df_clean = df.copy()

# Convert to lowercase
df_clean['Name_lower'] = df_clean['Name'].str.lower()

# Convert to uppercase
df_clean['City_upper'] = df_clean['City'].str.upper()

# String length
df_clean['Name_length'] = df_clean['Name'].str.len()

# Check if string contains substring
df_clean['Has_gmail'] = df_clean['Email'].str.contains('gmail', na=False)

# Extract domain from email
df_clean['Domain'] = df_clean['Email'].str.split('@').str[1]

print("String operations:")
print(df_clean[['Name', 'Name_lower', 'City_upper', 'Name_length', 'Has_gmail', 'Domain']])
print()

# Replace strings
df_clean['Email_masked'] = df_clean['Email'].str.replace('@', '_AT_')
print("Masked emails:")
print(df_clean[['Email', 'Email_masked']])
```

## 7. Data Type Conversion

```python
# Create sample data with mixed types
mixed_data = {
    'ID': ['1', '2', '3', '4', '5'],
    'Score': ['85.5', '90.0', '78.5', '92.0', '88.5'],
    'Grade': ['A', 'A', 'B', 'A', 'B'],
    'Passed': ['True', 'True', 'False', 'True', 'True']
}
df_mixed = pd.DataFrame(mixed_data)

print("Original data types:")
print(df_mixed.dtypes)
print()

# Convert data types
df_mixed['ID'] = pd.to_numeric(df_mixed['ID'])
df_mixed['Score'] = pd.to_numeric(df_mixed['Score'])
df_mixed['Grade'] = df_mixed['Grade'].astype('category')
df_mixed['Passed'] = df_mixed['Passed'].astype('bool')

print("After conversion:")
print(df_mixed.dtypes)
print()
print(df_mixed)
print()

# Handle conversion errors
problematic_data = pd.Series(['1', '2', 'three', '4', '5'])
converted = pd.to_numeric(problematic_data, errors='coerce')  # Convert errors to NaN
print("Handling conversion errors:")
print(converted)
```

## 8. Outlier Detection and Handling

```python
# Create data with outliers
np.random.seed(42)
normal_data = np.random.normal(50, 10, 100)
outliers = [150, 200, -50]  # Add some outliers
data_with_outliers = np.concatenate([normal_data, outliers])

df_outliers = pd.DataFrame({'Values': data_with_outliers})

print("Statistical summary:")
print(df_outliers.describe())
print()

# Detect outliers using IQR method
Q1 = df_outliers['Values'].quantile(0.25)
Q3 = df_outliers['Values'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_mask = (df_outliers['Values'] < lower_bound) | (df_outliers['Values'] > upper_bound)
print(f"Number of outliers: {outliers_mask.sum()}")
print("Outlier values:")
print(df_outliers[outliers_mask])
print()

# Remove outliers
df_no_outliers = df_outliers[~outliers_mask]
print(f"Original size: {len(df_outliers)}, After removing outliers: {len(df_no_outliers)}")
print("New statistical summary:")
print(df_no_outliers.describe())
```
