# Day 5: Working with Dates and Time Series

Today we'll learn how to work with dates, times, and time series data in pandas.

```python
import pandas as pd
import numpy as np
from datetime import datetime, date, timedelta
import matplotlib.pyplot as plt

# Display current date and time
print(f"Current datetime: {datetime.now()}")
print(f"Current date: {date.today()}")
```

## 1. Creating Date and Time Objects

```python
# Creating Timestamp objects
timestamp1 = pd.Timestamp('2024-01-15')
timestamp2 = pd.Timestamp('2024-01-15 14:30:00')
timestamp3 = pd.Timestamp(2024, 1, 15, 14, 30, 0)

print("Timestamp objects:")
print(f"Date only: {timestamp1}")
print(f"Date and time: {timestamp2}")
print(f"From components: {timestamp3}")
print()

# Creating date ranges
date_range1 = pd.date_range('2024-01-01', periods=10, freq='D')
date_range2 = pd.date_range('2024-01-01', '2024-01-10', freq='D')
date_range3 = pd.date_range('2024-01-01', periods=12, freq='M')

print("Date ranges:")
print(f"10 days: {date_range1}")
print(f"Between dates: {date_range2}")
print(f"Monthly: {date_range3}")
```

## 2. Parsing Dates from Strings

```python
# Sample data with date strings
date_strings = [
    '2024-01-15',
    '01/15/2024',
    '15-Jan-2024',
    '2024-01-15 14:30:00',
    'January 15, 2024'
]

# Parse different date formats
print("Parsing date strings:")
for date_str in date_strings:
    parsed = pd.to_datetime(date_str)
    print(f"'{date_str}' -> {parsed}")
print()

# Parse with specific format
custom_dates = ['15-01-2024', '16-01-2024', '17-01-2024']
parsed_custom = pd.to_datetime(custom_dates, format='%d-%m-%Y')
print("Custom format parsing:")
print(parsed_custom)
print()

# Handle parsing errors
problematic_dates = ['2024-01-15', 'not-a-date', '2024-02-30']
parsed_with_errors = pd.to_datetime(problematic_dates, errors='coerce')
print("Handling parsing errors:")
print(parsed_with_errors)
```

## 3. Working with DateTime Index

```python
# Create time series data
dates = pd.date_range('2024-01-01', periods=100, freq='D')
values = np.random.randn(100).cumsum()

ts = pd.Series(values, index=dates)
print("Time series data:")
print(ts.head(10))
print()

# Convert to DataFrame
df_ts = pd.DataFrame({
    'value': values,
    'abs_value': np.abs(values)
}, index=dates)

print("Time series DataFrame:")
print(df_ts.head())
print()

# Access datetime properties
print("DateTime properties:")
print(f"Year: {df_ts.index.year[:5].tolist()}")
print(f"Month: {df_ts.index.month[:5].tolist()}")
print(f"Day: {df_ts.index.day[:5].tolist()}")
print(f"Day of week: {df_ts.index.dayofweek[:5].tolist()}")
print(f"Day name: {df_ts.index.day_name()[:5].tolist()}")
```

## 4. Time-based Indexing and Selection

```python
# Select specific date
print("Value on 2024-01-15:")
print(ts['2024-01-15'])
print()

# Select date range
print("Values from Jan 10-15:")
print(ts['2024-01-10':'2024-01-15'])
print()

# Select by month
print("January 2024 values (first 10):")
print(ts['2024-01'].head(10))
print()

# Boolean indexing with dates
january_data = df_ts[df_ts.index.month == 1]
print(f"January data points: {len(january_data)}")
print(january_data.head())
print()

# Select weekdays only
weekdays = df_ts[df_ts.index.dayofweek < 5]
print(f"Weekday data points: {len(weekdays)}")
print(weekdays.head())
```

## 5. Resampling

```python
# Create hourly data
hourly_dates = pd.date_range('2024-01-01', periods=168, freq='H')  # 1 week
hourly_values = np.random.randn(168) * 10 + 100
hourly_ts = pd.Series(hourly_values, index=hourly_dates)

print("Hourly data (first 24 hours):")
print(hourly_ts.head(24))
print()

# Resample to daily (downsampling)
daily_mean = hourly_ts.resample('D').mean()
daily_sum = hourly_ts.resample('D').sum()
daily_max = hourly_ts.resample('D').max()

print("Daily resampling:")
print("Mean:", daily_mean)
print("Sum:", daily_sum)
print("Max:", daily_max)
print()

# Multiple aggregations
daily_agg = hourly_ts.resample('D').agg(['mean', 'std', 'min', 'max'])
print("Multiple aggregations:")
print(daily_agg)
print()

# Upsampling (fill missing values)
daily_data = pd.Series([100, 110, 105, 115], 
                      index=pd.date_range('2024-01-01', periods=4, freq='D'))
hourly_upsampled = daily_data.resample('H').ffill()  # Forward fill

print("Upsampling (first 25 hours):")
print(hourly_upsampled.head(25))
```

## 6. Time Zones

```python
# Create timezone-naive datetime
naive_ts = pd.Timestamp('2024-01-15 12:00:00')
print(f"Naive timestamp: {naive_ts}")
print(f"Timezone: {naive_ts.tz}")
print()

# Localize to timezone
utc_ts = naive_ts.tz_localize('UTC')
ny_ts = naive_ts.tz_localize('America/New_York')

print(f"UTC timestamp: {utc_ts}")
print(f"NY timestamp: {ny_ts}")
print()

# Convert between timezones
tokyo_ts = utc_ts.tz_convert('Asia/Tokyo')
london_ts = utc_ts.tz_convert('Europe/London')

print("Timezone conversions:")
print(f"UTC: {utc_ts}")
print(f"Tokyo: {tokyo_ts}")
print(f"London: {london_ts}")
print()

# Working with timezone-aware Series
dates_utc = pd.date_range('2024-01-01', periods=5, freq='D', tz='UTC')
ts_utc = pd.Series(range(5), index=dates_utc)

print("UTC Series:")
print(ts_utc)
print()

# Convert Series timezone
ts_tokyo = ts_utc.tz_convert('Asia/Tokyo')
print("Tokyo Series:")
print(ts_tokyo)
```

## 7. Date Arithmetic

```python
# Basic date arithmetic
base_date = pd.Timestamp('2024-01-15')

print("Date arithmetic:")
print(f"Base date: {base_date}")
print(f"Add 10 days: {base_date + pd.Timedelta(days=10)}")
print(f"Add 2 weeks: {base_date + pd.Timedelta(weeks=2)}")
print(f"Add 3 months: {base_date + pd.DateOffset(months=3)}")
print(f"Add 1 year: {base_date + pd.DateOffset(years=1)}")
print()

# Business day arithmetic
print("Business day arithmetic:")
print(f"Add 10 business days: {base_date + pd.BDay(10)}")
print(f"Next month end: {base_date + pd.MonthEnd()}")
print(f"Previous month begin: {base_date - pd.MonthBegin()}")
print()

# Calculate differences
date1 = pd.Timestamp('2024-01-01')
date2 = pd.Timestamp('2024-01-15')
diff = date2 - date1

print(f"Difference between {date2} and {date1}: {diff}")
print(f"Difference in days: {diff.days}")
print(f"Difference in seconds: {diff.total_seconds()}")
```

## 8. Working with Periods

```python
# Create periods
period_monthly = pd.Period('2024-01', freq='M')
period_quarterly = pd.Period('2024Q1', freq='Q')
period_yearly = pd.Period('2024', freq='Y')

print("Periods:")
print(f"Monthly: {period_monthly}")
print(f"Quarterly: {period_quarterly}")
print(f"Yearly: {period_yearly}")
print()

# Period ranges
monthly_periods = pd.period_range('2024-01', periods=12, freq='M')
quarterly_periods = pd.period_range('2024Q1', periods=4, freq='Q')

print("Period ranges:")
print(f"Monthly: {monthly_periods}")
print(f"Quarterly: {quarterly_periods}")
print()

# Convert between timestamps and periods
timestamps = pd.date_range('2024-01-01', periods=12, freq='M')
periods = timestamps.to_period('M')
back_to_timestamps = periods.to_timestamp()

print("Conversions:")
print(f"Timestamps: {timestamps[:3]}")
print(f"Periods: {periods[:3]}")
print(f"Back to timestamps: {back_to_timestamps[:3]}")
```

## 9. Time Series Analysis Example

```python
# Create sample stock price data
np.random.seed(42)
dates = pd.date_range('2023-01-01', '2024-01-01', freq='D')
prices = 100 + np.random.randn(len(dates)).cumsum()

stock_data = pd.DataFrame({
    'price': prices,
    'volume': np.random.randint(1000, 10000, len(dates))
}, index=dates)

print("Stock data sample:")
print(stock_data.head())
print()

# Calculate returns
stock_data['daily_return'] = stock_data['price'].pct_change()
stock_data['cumulative_return'] = (1 + stock_data['daily_return']).cumprod() - 1

# Rolling statistics
stock_data['price_ma_7'] = stock_data['price'].rolling(window=7).mean()
stock_data['price_ma_30'] = stock_data['price'].rolling(window=30).mean()
stock_data['volatility_30'] = stock_data['daily_return'].rolling(window=30).std()

print("With calculated metrics:")
print(stock_data[['price', 'daily_return', 'price_ma_7', 'price_ma_30']].head(10))
print()

# Monthly aggregation
monthly_stats = stock_data.resample('M').agg({
    'price': ['first', 'last', 'min', 'max'],
    'volume': 'sum',
    'daily_return': ['mean', 'std']
})

print("Monthly statistics:")
print(monthly_stats.head())
print()

# Quarterly performance
quarterly_return = stock_data['price'].resample('Q').apply(lambda x: (x.iloc[-1] / x.iloc[0]) - 1)
print("Quarterly returns:")
print(quarterly_return)
```

## 10. Handling Missing Dates

```python
# Create data with missing dates
irregular_dates = pd.to_datetime(['2024-01-01', '2024-01-03', '2024-01-07', '2024-01-10'])
irregular_data = pd.Series([100, 102, 98, 105], index=irregular_dates)

print("Irregular time series:")
print(irregular_data)
print()

# Reindex to fill missing dates
full_date_range = pd.date_range('2024-01-01', '2024-01-10', freq='D')
regular_data = irregular_data.reindex(full_date_range)

print("After reindexing (with NaN):")
print(regular_data)
print()

# Fill missing values
filled_data = regular_data.fillna(method='ffill')  # Forward fill
interpolated_data = regular_data.interpolate()     # Linear interpolation

print("Forward filled:")
print(filled_data)
print()

print("Interpolated:")
print(interpolated_data)
print()

# Business day calendar
business_days = pd.bdate_range('2024-01-01', '2024-01-15')
print("Business days only:")
print(business_days)
```

## Practice Exercises

```python
# Exercise 1: Sales data analysis
# Create daily sales data for 2023
np.random.seed(123)
sales_dates = pd.date_range('2023-01-01', '2023-12-31', freq='D')
daily_sales = np.random.normal(1000, 200, len(sales_dates))

# Add seasonal pattern (higher sales in December)
seasonal_boost = np.where(sales_dates.month == 12, 500, 0)
daily_sales += seasonal_boost

sales_df = pd.DataFrame({
    'sales': daily_sales
}, index=sales_dates)

# Monthly sales summary
monthly_sales = sales_df.resample('M').agg({
    'sales': ['sum', 'mean', 'std']
})
monthly_sales.columns = ['total_sales', 'avg_daily_sales', 'sales_std']

print("Monthly sales summary:")
print(monthly_sales.round(2))
print()

# Find best and worst performing months
best_month = monthly_sales['total_sales'].idxmax()
worst_month = monthly_sales['total_sales'].idxmin()

print(f"Best month: {best_month.strftime('%B %Y')} with ${monthly_sales.loc[best_month, 'total_sales']:,.2f}")
print(f"Worst month: {worst_month.strftime('%B %Y')} with ${monthly_sales.loc[worst_month, 'total_sales']:,.2f}")
print()

# Exercise 2: Calculate business metrics
# Add 7-day and 30-day moving averages
sales_df['ma_7'] = sales_df['sales'].rolling(window=7).mean()
sales_df['ma_30'] = sales_df['sales'].rolling(window=30).mean()

# Calculate year-over-year growth (if we had previous year data)
sales_df['day_of_year'] = sales_df.index.dayofyear
sales_df['month'] = sales_df.index.month

# Weekend vs weekday analysis
sales_df['is_weekend'] = sales_df.index.dayofweek >= 5
weekend_avg = sales_df[sales_df['is_weekend']]['sales'].mean()
weekday_avg = sales_df[~sales_df['is_weekend']]['sales'].mean()

print(f"Average weekend sales: ${weekend_avg:.2f}")
print(f"Average weekday sales: ${weekday_avg:.2f}")
print(f"Weekend vs Weekday ratio: {weekend_avg/weekday_avg:.2f}")
```
