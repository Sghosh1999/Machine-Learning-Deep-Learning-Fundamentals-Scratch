# Day 1: Data Selection and Indexing

Today we'll learn how to select, access, and manipulate data in pandas DataFrames and Series.

```python
import pandas as pd
import numpy as np

# Create sample data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, 30, 35, 28, 32],
    'City': ['New York', 'London', 'Tokyo', 'Paris', 'Berlin'],
    'Salary': [50000, 60000, 70000, 55000, 65000],
    'Department': ['IT', 'Finance', 'IT', 'HR', 'Finance']
}
df = pd.DataFrame(data)
print("Sample DataFrame:")
print(df)
```

## 1. Column Selection

```python
# Single column selection
print("Single column (Name):")
print(df['Name'])
print(type(df['Name']))  # Returns Series
print()

# Alternative syntax
print("Alternative syntax:")
print(df.Name)
print()

# Multiple columns
print("Multiple columns:")
print(df[['Name', 'Age', 'Salary']])
print(type(df[['Name', 'Age']]))  # Returns DataFrame
```

## 2. Row Selection

```python
# Using iloc (integer location)
print("First row using iloc:")
print(df.iloc[0])
print()

print("Multiple rows using iloc:")
print(df.iloc[0:3])
print()

# Using loc (label-based)
print("Row by index using loc:")
print(df.loc[1])
print()

print("Multiple rows using loc:")
print(df.loc[1:3])
```

## 3. Combined Selection (Rows and Columns)

```python
# iloc for specific rows and columns
print("Specific rows and columns with iloc:")
print(df.iloc[0:3, 1:4])  # Rows 0-2, Columns 1-3
print()

# loc for specific rows and columns
print("Specific rows and columns with loc:")
print(df.loc[0:2, 'Age':'Salary'])
print()

# Mixed selection
print("Mixed selection:")
print(df.loc[df['Age'] > 30, ['Name', 'Age', 'City']])
```

## 4. Boolean Indexing

```python
# Simple condition
print("People older than 30:")
print(df[df['Age'] > 30])
print()

# Multiple conditions with & (and)
print("People older than 25 AND in IT:")
print(df[(df['Age'] > 25) & (df['Department'] == 'IT')])
print()

# Multiple conditions with | (or)
print("People in IT OR Finance:")
print(df[(df['Department'] == 'IT') | (df['Department'] == 'Finance')])
print()

# Using isin() for multiple values
print("People in specific cities:")
print(df[df['City'].isin(['New York', 'London', 'Tokyo'])])
```

## 5. Adding and Removing Columns

```python
# Adding new columns
df['Bonus'] = df['Salary'] * 0.1
df['Total_Compensation'] = df['Salary'] + df['Bonus']
df['Age_Group'] = df['Age'].apply(lambda x: 'Young' if x < 30 else 'Senior')

print("DataFrame with new columns:")
print(df)
print()

# Removing columns
df_copy = df.copy()
df_copy = df_copy.drop(['Bonus'], axis=1)  # axis=1 for columns
print("After removing Bonus column:")
print(df_copy.columns.tolist())
print()

# Alternative way to remove columns
df_copy2 = df.copy()
del df_copy2['Total_Compensation']
print("After deleting Total_Compensation:")
print(df_copy2.columns.tolist())
```

## 6. Setting and Resetting Index

```python
# Set Name as index
df_indexed = df.set_index('Name')
print("DataFrame with Name as index:")
print(df_indexed)
print()

# Access by index name
print("Alice's data:")
print(df_indexed.loc['Alice'])
print()

# Reset index
df_reset = df_indexed.reset_index()
print("After resetting index:")
print(df_reset.head())
```

## 7. Data Types

```python
# Check data types
print("Data types:")
print(df.dtypes)
print()

# Convert data types
df['Age'] = df['Age'].astype('float64')
df['Department'] = df['Department'].astype('category')

print("After type conversion:")
print(df.dtypes)
print()

# Memory usage
print("Memory usage:")
print(df.memory_usage(deep=True))
```

## Practice Exercises

Try these exercises with the sample data:

```python
# Exercise 1: Select all employees with salary > 55000
high_earners = df[df['Salary'] > 55000]
print("High earners:")
print(high_earners[['Name', 'Salary']])
print()

# Exercise 2: Get the average age by department
avg_age_by_dept = df.groupby('Department')['Age'].mean()
print("Average age by department:")
print(avg_age_by_dept)
print()

# Exercise 3: Create a new column for salary category
df['Salary_Category'] = pd.cut(df['Salary'], 
                              bins=[0, 55000, 65000, float('inf')], 
                              labels=['Low', 'Medium', 'High'])
print("Salary categories:")
print(df[['Name', 'Salary', 'Salary_Category']])
```
