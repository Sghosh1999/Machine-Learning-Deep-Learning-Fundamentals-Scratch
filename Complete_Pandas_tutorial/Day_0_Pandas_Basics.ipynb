## Core Data Structures

### 1. Series (1D)
A Series is a one-dimensional array-like object containing data and an associated array of data labels (index).

```python
# Creating a Series
s1 = pd.Series([1, 2, 3, 4, 5])
print("Basic Series:")
print(s1)
print()

# Series with custom index
s2 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print("Series with custom index:")
print(s2)
print()

# Series from dictionary
data = {'apple': 5, 'banana': 3, 'orange': 8}
s3 = pd.Series(data)
print("Series from dictionary:")
print(s3)
```

### 2. DataFrame (2D)
A DataFrame is a 2D labeled data structure with columns of potentially different types.

```python
# Creating DataFrame from dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'London', 'Tokyo']
}
df = pd.DataFrame(data)
print("DataFrame from dictionary:")
print(df)
print()

# DataFrame info
print("DataFrame shape:", df.shape)
print("DataFrame columns:", df.columns.tolist())
print("DataFrame index:", df.index.tolist())
```

## Basic Operations

```python
# Head and tail
print("First 2 rows:")
print(df.head(2))
print()

print("Last 2 rows:")
print(df.tail(2))
print()

# Info about DataFrame
print("DataFrame info:")
df.info()
print()

# Describe
print("Statistical summary:")
print(df.describe())
```
