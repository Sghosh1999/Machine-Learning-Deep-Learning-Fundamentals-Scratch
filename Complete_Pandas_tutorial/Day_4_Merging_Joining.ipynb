# Day 4: Merging, Joining, and Reshaping Data

Today we'll learn how to combine multiple DataFrames and reshape data for analysis.

```python
import pandas as pd
import numpy as np

# Create sample datasets
employees = pd.DataFrame({
    'emp_id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'department': ['IT', 'Finance', 'IT', 'HR', 'Finance']
})

salaries = pd.DataFrame({
    'emp_id': [1, 2, 3, 4, 6],
    'salary': [50000, 60000, 70000, 55000, 65000],
    'bonus': [5000, 6000, 7000, 5500, 6500]
})

departments = pd.DataFrame({
    'department': ['IT', 'Finance', 'HR', 'Marketing'],
    'manager': ['John', 'Jane', 'Mike', 'Sarah'],
    'budget': [100000, 150000, 80000, 120000]
})

print("Employees DataFrame:")
print(employees)
print("\nSalaries DataFrame:")
print(salaries)
print("\nDepartments DataFrame:")
print(departments)
```

## 1. Inner Join

```python
# Inner join - only matching records
inner_join = pd.merge(employees, salaries, on='emp_id', how='inner')
print("Inner Join (employees + salaries):")
print(inner_join)
print(f"\nOriginal employees: {len(employees)}, Original salaries: {len(salaries)}, Inner join result: {len(inner_join)}")
```

## 2. Left Join

```python
# Left join - all records from left DataFrame
left_join = pd.merge(employees, salaries, on='emp_id', how='left')
print("Left Join (all employees):")
print(left_join)
print(f"\nResult length: {len(left_join)} (same as left DataFrame)")
```

## 3. Right Join

```python
# Right join - all records from right DataFrame
right_join = pd.merge(employees, salaries, on='emp_id', how='right')
print("Right Join (all salaries):")
print(right_join)
print(f"\nResult length: {len(right_join)} (same as right DataFrame)")
```

## 4. Outer Join

```python
# Outer join - all records from both DataFrames
outer_join = pd.merge(employees, salaries, on='emp_id', how='outer')
print("Outer Join (all records):")
print(outer_join)
print(f"\nResult length: {len(outer_join)}")
```

## 5. Merging on Different Column Names

```python
# Create DataFrames with different column names
df1 = pd.DataFrame({
    'employee_id': [1, 2, 3],
    'name': ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'emp_id': [1, 2, 3],
    'salary': [50000, 60000, 70000]
})

# Merge on different column names
merged_diff_cols = pd.merge(df1, df2, left_on='employee_id', right_on='emp_id')
print("Merging on different column names:")
print(merged_diff_cols)
print()

# Drop redundant column
merged_clean = merged_diff_cols.drop('emp_id', axis=1)
print("After dropping redundant column:")
print(merged_clean)
```

## 6. Merging on Multiple Columns

```python
# Create DataFrames with multiple key columns
sales_q1 = pd.DataFrame({
    'region': ['North', 'South', 'East', 'West'],
    'product': ['A', 'A', 'B', 'B'],
    'q1_sales': [100, 150, 200, 120]
})

sales_q2 = pd.DataFrame({
    'region': ['North', 'South', 'East', 'West'],
    'product': ['A', 'A', 'B', 'B'],
    'q2_sales': [110, 160, 180, 140]
})

# Merge on multiple columns
quarterly_sales = pd.merge(sales_q1, sales_q2, on=['region', 'product'])
print("Merging on multiple columns:")
print(quarterly_sales)
print()

# Calculate total sales
quarterly_sales['total_sales'] = quarterly_sales['q1_sales'] + quarterly_sales['q2_sales']
print("With total sales:")
print(quarterly_sales)
```

## 7. Index-based Joining

```python
# Create DataFrames with meaningful indices
emp_info = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35]
}, index=[1, 2, 3])

emp_salary = pd.DataFrame({
    'salary': [50000, 60000, 70000],
    'department': ['IT', 'Finance', 'IT']
}, index=[1, 2, 3])

# Join on index
joined_on_index = emp_info.join(emp_salary)
print("Joining on index:")
print(joined_on_index)
print()

# Alternative using merge with index
merged_on_index = pd.merge(emp_info, emp_salary, left_index=True, right_index=True)
print("Merging on index:")
print(merged_on_index)
```

## 8. Concatenation

```python
# Vertical concatenation (stacking rows)
df_2020 = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar'],
    'sales': [100, 120, 110],
    'year': [2020, 2020, 2020]
})

df_2021 = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar'],
    'sales': [110, 130, 125],
    'year': [2021, 2021, 2021]
})

# Concatenate vertically
vertical_concat = pd.concat([df_2020, df_2021], ignore_index=True)
print("Vertical concatenation:")
print(vertical_concat)
print()

# Horizontal concatenation (adding columns)
df_sales = pd.DataFrame({
    'month': ['Jan', 'Feb', 'Mar'],
    'sales': [100, 120, 110]
})

df_costs = pd.DataFrame({
    'costs': [80, 90, 85],
    'profit': [20, 30, 25]
})

# Concatenate horizontally
horizontal_concat = pd.concat([df_sales, df_costs], axis=1)
print("Horizontal concatenation:")
print(horizontal_concat)
print()

# Concatenation with keys
keyed_concat = pd.concat([df_2020, df_2021], keys=['2020', '2021'])
print("Concatenation with keys:")
print(keyed_concat)
```

## 9. Reshaping Data - Melt

```python
# Wide format data
wide_data = pd.DataFrame({
    'student': ['Alice', 'Bob', 'Charlie'],
    'math': [85, 90, 78],
    'science': [88, 85, 92],
    'english': [90, 88, 85]
})

print("Wide format data:")
print(wide_data)
print()

# Melt to long format
long_data = pd.melt(wide_data, id_vars=['student'], 
                   value_vars=['math', 'science', 'english'],
                   var_name='subject', value_name='score')

print("Long format data (melted):")
print(long_data)
print()

# Melt with multiple id variables
wide_data_2 = pd.DataFrame({
    'student': ['Alice', 'Bob', 'Charlie'],
    'grade': ['A', 'B', 'A'],
    'math': [85, 90, 78],
    'science': [88, 85, 92]
})

long_data_2 = pd.melt(wide_data_2, id_vars=['student', 'grade'],
                     var_name='subject', value_name='score')

print("Melt with multiple id variables:")
print(long_data_2)
```

## 10. Reshaping Data - Pivot

```python
# Convert long format back to wide format
pivoted_data = long_data.pivot(index='student', columns='subject', values='score')
print("Pivoted data (long to wide):")
print(pivoted_data)
print()

# Reset index to make student a column again
pivoted_reset = pivoted_data.reset_index()
print("After resetting index:")
print(pivoted_reset)
print()

# Rename columns index
pivoted_reset.columns.name = None
print("After removing columns name:")
print(pivoted_reset)
```

## 11. Stack and Unstack

```python
# Create MultiIndex DataFrame
arrays = [['A', 'A', 'B', 'B'], ['one', 'two', 'one', 'two']]
index = pd.MultiIndex.from_arrays(arrays, names=['first', 'second'])
df_multi = pd.DataFrame({'values': [1, 2, 3, 4]}, index=index)

print("MultiIndex DataFrame:")
print(df_multi)
print()

# Unstack - move index level to columns
unstacked = df_multi.unstack()
print("Unstacked:")
print(unstacked)
print()

# Stack - move columns to index
stacked = unstacked.stack()
print("Stacked back:")
print(stacked)
print()

# Working with regular DataFrame
df_regular = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6],
    'C': [7, 8, 9]
}, index=['X', 'Y', 'Z'])

print("Regular DataFrame:")
print(df_regular)
print()

# Stack columns to create Series
stacked_regular = df_regular.stack()
print("Stacked regular DataFrame:")
print(stacked_regular)
print()

# Unstack back to DataFrame
unstacked_regular = stacked_regular.unstack()
print("Unstacked back:")
print(unstacked_regular)
```

## 12. Complex Merging Example

```python
# Combine all our sample data
# First merge employees with salaries
emp_sal = pd.merge(employees, salaries, on='emp_id', how='left')
print("Employees + Salaries:")
print(emp_sal)
print()

# Then merge with departments
complete_data = pd.merge(emp_sal, departments, on='department', how='left')
print("Complete employee data:")
print(complete_data)
print()

# Calculate some metrics
complete_data['total_compensation'] = complete_data['salary'].fillna(0) + complete_data['bonus'].fillna(0)
complete_data['budget_per_employee'] = complete_data.groupby('department')['budget'].transform(
    lambda x: x.iloc[0] / len(x) if len(x) > 0 else 0
)

print("With calculated metrics:")
print(complete_data[['name', 'department', 'total_compensation', 'budget_per_employee']])
```

## Practice Exercises

```python
# Exercise 1: Create and merge customer and order data
customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['John', 'Jane', 'Bob', 'Alice'],
    'city': ['NY', 'LA', 'Chicago', 'Boston']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105],
    'customer_id': [1, 2, 1, 3, 2],
    'amount': [100, 200, 150, 300, 250]
})

# Merge and calculate total orders per customer
customer_orders = pd.merge(customers, orders, on='customer_id', how='left')
order_summary = customer_orders.groupby(['customer_id', 'name', 'city']).agg({
    'order_id': 'count',
    'amount': 'sum'
}).rename(columns={'order_id': 'total_orders', 'amount': 'total_amount'})

print("Customer order summary:")
print(order_summary.reset_index())
print()

# Exercise 2: Reshape sales data
monthly_sales = pd.DataFrame({
    'product': ['A', 'B', 'C'],
    'jan': [100, 150, 200],
    'feb': [110, 160, 180],
    'mar': [120, 140, 220]
})

# Melt to long format
sales_long = pd.melt(monthly_sales, id_vars=['product'], 
                    var_name='month', value_name='sales')

# Calculate total sales by month
monthly_totals = sales_long.groupby('month')['sales'].sum()

print("Monthly sales totals:")
print(monthly_totals)
```
