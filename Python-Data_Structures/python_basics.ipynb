```python
## Matrix Multiplication

matrix_a = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]]

matrix_b = [
    [9, 8],
    [6, 5],
    [3, 2]]

def matrix_multiply(a, b):
    row1, col1 = len(a), len(a[0])
    row2, col2 = len(b), len(b[0])

    if col1 != row2:
        raise ValueError("No of columns in Matrix 1 must be equal to No of Columns in Matrix2")

    result = [[0] * row2 for col in range(col1)]

    for i in range(row1):
        for j in range(col2):
            for k in range(col1):
                result[i][k] = a[i][k] * b[k][j]
    
    print(result)
    
```

```python
matrix_multiply(matrix_a, matrix_b)
```
